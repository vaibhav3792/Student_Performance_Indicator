## Student Performance Indicator


### 1) Problem statement
- This project understands how the student's performance (test scores) is affected by other variables such as Gender, Ethnicity, Parental level of education, Lunch and Test preparation course.


### 2) Data Collection
- Dataset Source - https://www.kaggle.com/datasets/spscientist/students-performance-in-exams?datasetId=74977
- The data consists of 8 column and 1000 rows.

### Adding columns for "Total Score" and "Average"

```python
df['total score'] = df['math_score'] + df['reading_score'] + df['writing_score']
df['average'] = df['total score']/3
df.head()
```

```python
reading_full = df[df['reading_score'] == 100]['average'].count()
writing_full = df[df['writing_score'] == 100]['average'].count()
math_full = df[df['math_score'] == 100]['average'].count()

print(f'Number of students with full marks in Maths: {math_full}')
print(f'Number of students with full marks in Writing: {writing_full}')
print(f'Number of students with full marks in Reading: {reading_full}')
```

```python
reading_less_20 = df[df['reading_score'] <= 20]['average'].count()
writing_less_20 = df[df['writing_score'] <= 20]['average'].count()
math_less_20 = df[df['math_score'] <= 20]['average'].count()

print(f'Number of students with less than 20 marks in Maths: {math_less_20}')
print(f'Number of students with less than 20 marks in Writing: {writing_less_20}')
print(f'Number of students with less than 20 marks in Reading: {reading_less_20}')
```

#####  Insights
 - From above values we get students have performed the worst in Maths
 - Best performance is in reading section

#### BIVARIATE ANALYSIS ( Does gender have any impact on student's performance ? )

```python
gender_group = df.groupby('gender').mean(numeric_only=True)
gender_group
```

```python
plt.figure(figsize=(10, 8))

X = ['Total Average','Math Average']


female_scores = [gender_group['average'][0], gender_group['math_score'][0]]
male_scores = [gender_group['average'][1], gender_group['math_score'][1]]

X_axis = np.arange(len(X))

plt.bar(X_axis - 0.2, male_scores, 0.4, label = 'Male')
plt.bar(X_axis + 0.2, female_scores, 0.4, label = 'Female')

plt.xticks(X_axis, X)
plt.ylabel("Marks")
plt.title("Total average v/s Math average marks of both the genders", fontweight='bold')
plt.legend()
plt.show()
```

#### Insights
- On an average females have a better overall score than men.

#### BIVARIATE ANALYSIS (  Does lunch type intake have any impact on student's performance ? )

```python
plt.style.use('seaborn-v0_8-whitegrid')

fig, ax = plt.subplots(2, 2, figsize=(14, 10))

plots = [
    ('math_score', 'Math Score'),
    ('reading_score', 'Reading Score'),
    ('writing_score', 'Writing Score'),
    ('average', 'Average Score')
]

for axis, (col, title) in zip(ax.flatten(), plots):
    sns.barplot(
        x='lunch',
        y=col,
        data=df,
        ax=axis,
        palette='bright'
    )

    axis.set_title(title, fontsize=16, fontweight='bold')
    axis.set_xlabel('Lunch Type', fontsize=12)
    axis.set_ylabel('Score', fontsize=12)


    for container in axis.containers:
        axis.bar_label(container, fmt='%.1f', fontsize=11)

plt.suptitle('Impact of Lunch Type on Student Performance',
             fontsize=20, fontweight='bold')

plt.tight_layout()
plt.show()
```

#### Insights
- Students who get Standard Lunch tend to perform better than students who got free/reduced lunch

####TEST PREPARATION COURSE COLUMN
- Do students complete the test preparation course?
- Does Test prepration course have any impact on student's performance ?

#### UNIVARIATE ANALYSIS (Do the students take test preparation course?)

```python
plt.figure(figsize=(7,5))
sns.countplot(x='test_preparation_course', data=df, palette='bright')

plt.title('Distribution of Test Preparation Course', fontsize=14, fontweight='bold')
plt.xlabel('Test Preparation Course')
plt.ylabel('Number of Students')

plt.show()
```

Most students did not complete the test preparation course.

#### BIVARIATE ANALYSIS ( Does Test prepration course have any impact on student's performance ? )

```python
plt.style.use('seaborn-v0_8-whitegrid')

fig, ax = plt.subplots(1, 3, figsize=(15,5))

subjects = ['math_score', 'reading_score', 'writing_score']

for axis, subject in zip(ax, subjects):
    sns.barplot(
        x='test_preparation_course',
        y=subject,
        data=df,
        palette='bright',
        ax=axis
    )

    axis.set_title(subject.replace('_', ' ').title(), fontsize=12, fontweight='bold')
    axis.set_xlabel('')
    axis.set_ylabel('Score')

plt.suptitle('Average Scores by Test Preparation Course',
             fontsize=16, fontweight='bold')

plt.tight_layout()
plt.show()
```

#### Insights  
- Students who have completed the Test Prepration Course have scores higher in all three categories than those who haven't taken the course

```python
df['average_score'] = df[['math_score','reading_score','writing_score']].mean(axis=1)

plt.figure(figsize=(7,5))
sns.boxplot(x='test_preparation_course', y='average_score', data=df, palette='bright')

plt.title('Overall Performance by Test Preparation Course', fontweight='bold')
plt.show()
```
